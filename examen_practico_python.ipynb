### Ejercicio 1: Resultado de un partido de fútbol (if-elif-else)
**Enunciado:** Escribe una función llamada `resultado_partido(goles_local, goles_visitante)` que reciba el número de goles de dos equipos. Si `goles_local > goles_visitante`, debe devolver `"Gana local"`. Si `goles_local < goles_visitante`, debe devolver `"Gana visitante"`. Si ambos goles son iguales, debe devolver `"Empate"`.

```python
# --- Escribe aquí tu solución ---
def resultado_partido(goles_local, goles_visitante):

    goles_local = list([3, 0, 2])
    goles_visitante = list([1, 2, 2])
    if goles_local > goles_visitante:
            return("Gana local")
    elif goles_local == goles_visitante:
            return("Empate")
    else:
            return("Gana visitante")

```

```python
# --- Test Unitario (No modificar) ---
assert resultado_partido(3, 1) == "Gana local"
assert resultado_partido(0, 2) == "Gana visitante"
assert resultado_partido(2, 2) == "Empate"
print('✅ Ejercicio 1 (Fútbol) superado con éxito!')
```
